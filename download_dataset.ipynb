```python
import os
from kaggle.api.kaggle_api_extended import KaggleApi
import zipfile

ROOT = os.getcwd()
DATASET_DIR = os.path.join(ROOT, "dataset")

os.makedirs(DATASET_DIR, exist_ok=True)

api = KaggleApi()
api.authenticate()

print("⬇ Downloading dataset...")
api.dataset_download_files(
    "smaranjitghose/corn-or-maize-leaf-disease-dataset",
    path=DATASET_DIR,
    unzip=False
)
print("✅ Dataset ZIP downloaded!")

# unzip manual via python
zip_path = os.path.join(DATASET_DIR, "corn-or-maize-leaf-disease-dataset.zip")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(DATASET_DIR)

print("📁 Dataset extracted!")
```

```python
import os, shutil

DATASET_DIR = "E:\semester7\comvis\deteksidaun\dataset\data"

mapping = {
    "Healthy": "Daun Sehat",
    "Blight": "Hawar Daun",
    "Common_Rust": "Karat Daun",
    "Gray_Leaf_Spot": "Bercak Daun"
}

for old, new in mapping.items():
    old_path = os.path.join(DATASET_DIR, old)
    new_path = os.path.join(DATASET_DIR, new)
    if os.path.exists(old_path):
        shutil.move(old_path, new_path)
        print(f"📁 {old} → {new}")
    else:
        print(f"⚠ {old} tidak ditemukan")

print("✅ Renaming selesai!")
```
