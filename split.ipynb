```python
import os, random, shutil
from pathlib import Path

SRC = Path(r"E:\semester7\comvis\deteksidaun\dataset\data")     # <- folder data mentah (4 kelas)
OUT = Path(r"E:\semester7\comvis\deteksidaun\dataset\split")   # <- output split

train_ratio, val_ratio, test_ratio = 0.70, 0.15, 0.15
seed = 42
random.seed(seed)

IMG_EXT = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def is_img(p: Path):
    return p.suffix.lower() in IMG_EXT

# bersihkan split lama kalau ada
if OUT.exists():
    shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)

classes = [d.name for d in SRC.iterdir() if d.is_dir()]
print("Classes:", classes)

for cls in classes:
    imgs = [p for p in (SRC/cls).iterdir() if p.is_file() and is_img(p)]
    random.shuffle(imgs)

    n = len(imgs)
    n_train = int(train_ratio * n)
    n_val   = int(val_ratio * n)
    n_test  = n - n_train - n_val

    splits = {
        "train": imgs[:n_train],
        "val":   imgs[n_train:n_train+n_val],
        "test":  imgs[n_train+n_val:]
    }

    for sp, files in splits.items():
        dst = OUT/sp/cls
        dst.mkdir(parents=True, exist_ok=True)
        for f in files:
            shutil.copy2(f, dst/f.name)

    print(f"{cls}: total={n} train={n_train} val={n_val} test={n_test}")

print("✅ Split selesai:", OUT)
```

```python
import os

ROOT = r"E:/semester7/comvis/CornLeaf_Disease/dataset/split"
splits = ["train", "val", "test"]

def count_images(path):
    IMG_EXT = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
    return sum(
        1 for f in os.listdir(path)
        if f.lower().endswith(IMG_EXT)
    )

for sp in splits:
    print(f"\n=== {sp.upper()} ===")
    sp_dir = os.path.join(ROOT, sp)
    total = 0
    for cls in sorted(os.listdir(sp_dir)):
        cls_dir = os.path.join(sp_dir, cls)
        if not os.path.isdir(cls_dir): 
            continue
        n = count_images(cls_dir)
        total += n
        print(f"{cls:12s}: {n}")
    print("TOTAL:", total)
```
